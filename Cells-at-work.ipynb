```python
data1=pd.read_csv(root/"bilibilib_gongzuoxibao.csv")
fig,axes=plt.subplots(3,3)

i=0
j=0
for col in ["score","disliked","likes","liked","last_ep_index","date"]:
    sns.histplot(data1,x=col,ax=axes[i][j])
    if j<2 and i<2:
        j+=1
    elif j==2 and i<2: 
        i+=1
        j=0
    else:
        break
```
